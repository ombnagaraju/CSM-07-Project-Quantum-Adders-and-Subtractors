# Final Experimental Result

The optimized reversible quantum adder-cum-subtractor was
implemented and evaluated using Qiskit.

Compared with the baseline implementation, the optimized
circuit reduced gate count from 9 to 6, circuit depth from
7 to 4, and quantum cost from 21 to 10.

The circuit was also extended to 8-bit arithmetic and verified
using multiple test cases.

For image processing, two 16×16 grayscale images were added
pixel-by-pixel using the quantum arithmetic implementation.
The resulting image was then subtracted by the second input
image to recover the first image.

The recovered image showed zero pixel difference from the
original image. The measured image-recovery results were
PSNR = ∞ and SSIM = 1.0.
