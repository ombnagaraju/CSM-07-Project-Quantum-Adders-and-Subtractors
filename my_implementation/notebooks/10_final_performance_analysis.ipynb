```python
import pandas as pd
import numpy as np

# Confirmed measured circuit results

performance = pd.DataFrame({
    "Metric": [
        "Number of Qubits",
        "Gate Count",
        "CNOT Count",
        "Toffoli Count",
        "Circuit Depth",
        "Quantum Cost"
    ],

    "Baseline": [
        5,
        9,
        6,
        3,
        7,
        21
    ],

    "Optimized": [
        5,
        6,
        5,
        1,
        4,
        10
    ]
})

print(performance)
```

```python
improvements = {
    "Gate Count Reduction (%)": ((9 - 6) / 9) * 100,
    "CNOT Reduction (%)": ((6 - 5) / 6) * 100,
    "Toffoli Reduction (%)": ((3 - 1) / 3) * 100,
    "Circuit Depth Reduction (%)": ((7 - 4) / 7) * 100,
    "Quantum Cost Reduction (%)": ((21 - 10) / 21) * 100
}

improvement_df = pd.DataFrame(
    list(improvements.items()),
    columns=["Metric", "Improvement (%)"]
)

print(improvement_df)
```

```python
performance.to_csv(
    "../results/tables/final_circuit_performance_comparison.csv",
    index=False
)

improvement_df.to_csv(
    "../results/tables/final_circuit_improvements.csv",
    index=False
)

print("Final circuit performance results saved.")
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
metrics = ["Gate Count", "Circuit Depth", "Quantum Cost"]

baseline_values = [9, 7, 21]
optimized_values = [6, 4, 10]

x = np.arange(len(metrics))
width = 0.35

plt.figure(figsize=(9, 5))

plt.bar(
    x - width/2,
    baseline_values,
    width,
    label="Baseline"
)

plt.bar(
    x + width/2,
    optimized_values,
    width,
    label="Optimized"
)

plt.xlabel("Performance Metrics")
plt.ylabel("Value")
plt.title("Baseline vs Optimized Adder-Cum-Subtractor")
plt.xticks(x, metrics)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
reductions = [33.33, 42.86, 52.38]

for i, reduction in enumerate(reductions):
    print(f"{metrics[i]} reduction: {reduction:.2f}%")
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    x - width/2,
    baseline_values,
    width,
    label="Baseline"
)

plt.bar(
    x + width/2,
    optimized_values,
    width,
    label="Optimized"
)

plt.xlabel("Performance Metrics")
plt.ylabel("Value")
plt.title("Baseline vs Optimized Adder-Cum-Subtractor")
plt.xticks(x, metrics)
plt.legend()

plt.tight_layout()

plt.savefig(
    "../results/circuits/baseline_vs_optimized_performance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Performance comparison graph saved.")
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    metrics,
    reductions
)

plt.xlabel("Performance Metrics")
plt.ylabel("Reduction (%)")
plt.title("Resource Reduction After Optimization")

for i, value in enumerate(reductions):
    plt.text(
        i,
        value + 1,
        f"{value:.2f}%",
        ha="center"
    )

plt.ylim(0, 60)
plt.tight_layout()

plt.savefig(
    "../results/circuits/optimization_reduction.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Optimization reduction graph saved.")
```
