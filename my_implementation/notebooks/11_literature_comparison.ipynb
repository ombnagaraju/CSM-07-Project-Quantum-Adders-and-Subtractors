## Research Direction

The selected studies demonstrate reversible arithmetic design,
adder-subtractor architectures, quantum-circuit optimization,
and image-processing applications using different technologies.

However, the approaches use different circuit structures,
technologies and performance measures.

Our work focuses on developing a single optimized reversible
quantum adder-cum-subtractor using Qiskit and evaluating it
using a consistent set of circuit-resource metrics.

The proposed implementation is further applied to image
addition and image recovery through subtraction, followed by
PSNR and SSIM evaluation.
